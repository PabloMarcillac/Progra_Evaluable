A continuación te muestro el uso de turtle. Lee los comentarios.

```python
#importo la biblioteca de turtle
import turtle
import math

# definimos una función arc que dibuja un arco de circunferencia 
# con un radio (r) y un ángulo determinados, a partir de la posición actual de la tortuga.
def arc(t, r, ángulo):
    longitud_del_arco = 2 * math.pi * r * ángulo / 360
    n = int(longitud_del_arco / 3) + 1
    longitud_de_paso = longitud_del_arco / n
    angulo_de_paso = ángulo / n
    
    for i in range(n):
        t.fd(longitud_de_paso)
        t.lt(angulo_de_paso)

# definimos una función petalo que permite crear un petalo a partir de la función arc, creando dos arcos simétricos
def petalo(t, r, angulo):
    for i in range(2):
        arc(t, r, angulo)
        t.lt(abs(180-angulo))

#definimos una función flor que permite crear un flor a partir de la función petalo
def flor(t, n, r, angulo):
    for i in range(n):
        petalo(t, r, angulo)
        t.lt(360/n)
       

#defino la tortuga que voy a utilizar
t = turtle.Turtle()
t.speed(0)
# Dibujo una flor desde la posición actual de la tortuga
flor(t, 6, 300, 60)

# Finalizo la ventana del dibujo permitiendo que se cierre con un click de ratón para que no haya problemas.
turtle.exitonclick()
```

Escribe un conjunto de funciones apropiadamente general que pueda dibujar polígonos regulares con segmentos que unen el centro del polígono con cada uno de los vértices. Ahora, los pasos que antes te indiqué yo tienes que definirlos tú para este nuevo problema. Hazlo antes de lanzarte a resolverlo.

```python
import turtle
import math

#Defino una función polígono que permite dibujar un polígono de n lados
def poligono(t, longitud, n):
    for _ in range(n):
        t.fd(longitud)
        t.lt(360 / n)

#Defino una función centro que calcula la distancia del centro del polígono al vértice
def centro(longitud, n):
    return (longitud / (2 * math.sin(math.pi / n)))

#Defino una función segmentos que dibuja segmentos que unen el centro con cada vértice
def segmentos(t, longitud, n):
    c=centro(longitud, n)
    angulo=360/n
    for _ in range(n):
        t.fd(c)      # dibuja desde el centro hacia el vértice
        t.bk(c)      # vuelve al centro
        t.lt(angulo)

#Defino una función figura que permite dibujar el polígono con segmentos que unen el centro del polígono a cada vértice
def figura(t, longitud, n):
    c=centro(longitud, n)
    # Ir al centro del polígono
    t.penup()
    t.goto(0, 0) #mueve la tortuga directamente a las coordenadas pedidas (en este caso x=0, y=0) 
    t.pendown()
    # Dibuja los segmentos primero (desde el centro)
    segmentos(t, longitud, n)
    # Coloca la tortuga en el primer vértice y dibuja el polígono
    t.penup()
    t.setheading(0) #cambia la orientación de la tortuga (0 se orienta a la derecha)
    t.fd(c)
    t.lt(90 + 180/n)
    t.pendown()
    poligono(t, longitud, n)

# Crear tortuga
t = turtle.Turtle()
t.speed(0)

# Dibujar figura
figura(t, 100, 8)

# Finalizo la ventana del dibujo permitiendo que se cierre con un click de ratón para que no haya problemas.
turtle.exitonclick()
```

Lee sobre espirales en http://en.wikipedia.org/wiki/Spiral; luego escribe un programa que dibuje una espiral arquimediana (o de otro tipo). Al igua que antes debes definir preciamente las funciones, qué hace y cómo vas a irlas probando.

```python
import turtle
import math

#Defino una función espiral que calcula la posición en coordenadas de la espiral de Arquímedes, con el ángulo theta (en radianes) y los parámetros a y b
#(las fórmulas están sacadas de Wikipedia)
def punto_espiral(theta, a, b):
    r = a + b * theta
    x = r * math.cos(theta)
    y = r * math.sin(theta)
    return x, y

#Defino una función espiral que dibuja la espiral de Arquímedes a partir de los parámetros a y b, desde theta=0 hasta theta=theta_max, avanzando en incrementos
def espiral(t, a, b, theta_max, incremento):
    theta = 0
    prim_posición=True
    while theta<=theta_max:
        x, y = punto_espiral(theta, a, b) #calcular la posición (x, y) en la espiral
    if prim_posición:
        t.penup()
        t.goto(x, y)                      #mover tortuga a la posición inicial
        t.pendown()
        prim_posición=False               #marcar que ya estamos en la primera posición
    else:
        t.goto(x, y)                      #dibujar hasta la siguiente posición
    theta+=incremento                     #avanzar el ángulo para el siguiente punto

#Crear tortuga
t=turtle.Turtle()
t.speed(0)
vueltas=5                                 #número de vueltas que quiero dibujar
theta_max=2*math.pi*vueltas               #ángulo máximo en radianes 
#Dibujar espiral
espiral(t, 0, 5, theta_max, 0.04)

# Finalizo la ventana del dibujo permitiendo que se cierre con un click de ratón para que no haya problemas.
turtle.exitonclick()
```

Busca información sobre el fractal de Koch y reprodúcelo. Sugerencia: realiza una aproximación recursiva con un límite de recursividad. Simplifica haciendo que la longitud inicial de los lados tenga que ser una potencia de 3. El límite de recursividad es cuando la longitud del segmento a dibujar es de tamaño 1 (tal vez con tu algoritmo, depende de cómo lo implementes, debas hacerlo cuando sea menor que tres)

```python
import turtle

#Defino una función potencia3 que comprueba si n es potencia de 3
def potencia3(n):
    if n <= 0:
        return False
    while n % 3 == 0:
        n //= 3
    return n == 1

#Defino una función koch que dibuja un segmento del fractal de Koch con n iteraciones y longitud de segmento(longitud)  
def koch(t, longitud, n):
    #si n==0, solo avanza longitud
    if n == 0:
        t.fd(longitud)
    #si n>0, divide en 4 segmentos de L con los giros correspondientes
    else:
        L= longitud/3
        #Forma el primer lado del triángulo
        koch(t, L, n - 1)
        t.lt(60)
        #Gira 120° a la derecha y dibuja el siguiente segmento, esto baja la tortuga hacia el otro lado del triángulo
        koch(t, L, n - 1)
        t.rt(120)
        koch(t, L, n - 1)
        #Gira otra vez 60° a la izquierda para recuperar la orientación inicial y dibuja el último tercio
        t.lt(60)
        koch(t, L, n - 1)
        
#Defino una función fractal que dibuja el fractal de Koch completo usando la longitud y n como número de iteraciones
def fractal(t, longitud, n):
    for _ in range(3):
        koch(t, longitud, n)
        t.rt(120)

#Crear tortuga
t=turtle.Turtle()
t.speed(0)

#Dibujar fractal
fractal(t, 243, 4)

# Finalizo la ventana del dibujo permitiendo que se cierre con un click de ratón para que no haya problemas.
turtle.exitonclick()
```
